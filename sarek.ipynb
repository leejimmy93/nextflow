### Set up ~/.nextflow/config or a local nextflow.config

```python
%%sh
# Create directory in tmp for user
mkdir -p /tmp/$USER
 


# Write nextflow config file to that directory
echo """process {
    executor = 'awsbatch'
    queue = 'nextflow-base-v1'
    
}
aws {
    region = 'us-east-1'
    batch {
        cliPath = '/root/miniconda/bin/aws'
        jobRole = 'arn:aws:iam::335777049998:role/ecsTaskExecutionRole'
    }
}""" > /tmp/$USER/config
      
```
